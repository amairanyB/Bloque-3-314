#### Estructura if
if permite ejecutar instrucciones únicamente cuando se cumple una condición </br>
-  if_ evalúa una condición
- .isnull(): identifica los valores nulos
- any(): devuelve True si al menos un valor cumple la condición

```python
# verificar si existen valores nulos, se comprueba si la columna contiene al menos un valor nulo y, si es así, se muestra un mensaje.
if colores_especializaciones['id_distrito_judicial'].isnull().any(): # any comprueba si al menos uno de los valores es verdadero
    print("Existen valores nulos en la columna id_distrito_judicial")
```

```python
# identificar si existen registros de tipo de elección 8
if(colores_especializaciones['id_tipo_eleccion'] == 8).any():
    print("Existen registros del tipo de eleccion 8")
```

#### Estructura if else
permite ejecutar una instrucción cuando la condicion se cumple y otra cuando no se cumple
-  .empty: comprueba si un Datframe esta vacío
-  not: invierte el resultado de una condición
- any(): devuelve True si al menos un valor cumple la condición

```python
# comprobar si existen registros con el lado izquierdo (I)
if(colores_especializaciones['lado'] == 'I').any(): # devuelve True e imprime si al menos un valor es I, PUEDES USAR i PARA EJEMPLIFICAR    
    print("Existen registros del lado izquierdo")
else:
    print("No existen registros del aldo izquierdo")
```

```python
# Comprobar si el Dataframe contiene registros
if not colores_especializaciones.empty:
    print("El datafame contiene registros")
else:
    print("El dataframe está vacío")
```

#### Estructura if - elif - else
Permite evaluar condiciones. Si la primera no se cumple, comprueba la siguiente y si ninguna se cumple, ejecuta else

```python
# clasificar los registros según el lado al que pertenecen
# la columna lado contiene I, D y C

# obtener por registro: 0 es I, 61 es C y 79 es D
lado = colores_especializaciones.loc[79, 'lado']

if lado == 'I':
    print("El registro pertenece al lado izquierdo")
elif lado == 'D':
    print("El registro pertenece al lado derecho")
elif lado == 'C':
    print("El registro pertenece al centro")
else:
    print("Lado no identificado")
```

```python
# clasificar los registros según su tipo de elección, por ejemplo el 0 o 33
tipo = colores_especializaciones.loc[33, 'id_tipo_eleccion']

if tipo == 7:
    print("Tipo de eleccion 7")
elif tipo == 8:
    print("Tipo elección 8")
else:
    print("otro tipo de elección")
```

#### Ciclo for
permite recorrer registros de un dataframe
 - for: recorre los valores de la columna
 - enumerate(): proporciona el índice y el valor de cada elemento
 - pd.isnull(): comprueba si el valor es nulo
 - enumerate(): permite recorrer los elementos de una colección y obtener dos datos, su posición y su valor

```python
# recorrer e imprimir todos los colores de la columna color
for color in colores_especializaciones['color']:
    print(color)
```

```python
# recorrer los colores e IMPRIMIR cuando se encuentren el código hexadecimal #ECEDF5
for color in colores_especializaciones['color']:
    if color == '#ECEDF5':
        print("Se encontro el color #ECEDF5")
```

```python
# recorrer los registros e identificar los que no tienen distrito judicial asignado
# idice: variable que almacena la posición del registro dentro del DataFrame: 0, 1, 2, 3 ... que recibe d ela funcion enumerate
# distrito: variable que almacena el valor de id_distrito_judicial correspondiente a cada registro
for indice, distrito in enumerate(colores_especializaciones['id_distrito_judicial']):
    if pd.isnull(distrito):
        print("Registro sin distrito judicial: ", indice)
#Se recorren todos los registros de la columna id_distrito_judicial del DataFrame colores_especializaciones. 
# Mediante enumerate se obtiene tanto la posición de cada registro como su valor. Después, se verifica si el valor del distrito judicial es nulo 
# y, si lo es, se muestra la posición del registro que no tiene un distrito judicial asignado.
```

```python
# recorrer los registros y clasificarlos según su lado
# puedes comentar un elif para obtener el 'sin clasificacion'
for lado in colores_especializaciones['lado']:
    if lado == 'I':
        print("Izquiero")
    elif lado == 'D':
        print("Derecho")
    elif lado == 'C':
        print("Centro")
    else:
        print("Sin clasificación")
```

#### ciclo while
ejecuta instrucciones mientras una condición sea verdadera. Requiere contadores
- len(): obtiene la cantidad de registros del DataFrame
- break: termina el ciclo cuando se encuentra el color buscado

```python
# recorrer e imprimir los 10 primeros colores del dataframe

# inicializar contador
i = 0

# recorrer mientras el contador sea menor a 10
while i < 10:
    print(colores_especializaciones.loc[i, 'color'])
    # incrementar contador
    i += 1
```

```python
# recorrer los registros hasta encontrar un color específico
# #f7e9f0 esta en el 3
# #86b1d7 esta en el 5
# #DCCCC6 esta en el 30

# inicializar contador
i = 0

# recorrer mientras no se encuentre el color
while i < len(colores_especializaciones):
    if colores_especializaciones.loc[i, 'color'] == '#f7e9f0':
        print("Color encontrado en el registro: ", i)
        break
    # aumentar el contador
    i += 1
```
