```python
# librerías
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# importar la clase de conexion a la BD
from conexionBD import conexionBD
# crear el objeto de la clase para usarla
bd = conexionBD()
```

```python
# consulta a la BD para obtener nuestro dataframe de la tabla tcolores_especializaciones
consulta = """ SELECT * FROM siac.tcolores_especializaciones"""
```

```python
# obtener nuestro Dataframe
tablaColores = pd.read_sql(consulta, bd.conexion)
# mostrar dataframe
tablaColores
```

#### Visualización usando MATPLOTLIB y SEABORN 

```python
# Crear un gráfico de conteo (frecuencias)
sns.countplot(
    data = tablaColores,
    x = 'color',
    color = '#86b1d7'
)

# Opcional: rotar las etiquetas del eje X si los códigos de color se enciman
plt.xticks(rotation=45)

plt.show()
```
